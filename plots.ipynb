```python
import networkx as nx
from matplotlib import pyplot as plt
import pickle
#reload 
%load_ext autoreload
%autoreload 2
from chart_utils import make_plots
import numpy as np

def re_title(env_name):
    if env_name == 'four_rooms_11_11.data':
        env_name = 'Four Rooms'
    if env_name == 'office':
        env_name = 'Office'
    if env_name == 'toh':
        env_name = 'Tower of Hanoi'
    if env_name == 'torus':
        env_name = 'Torus (Directed)'
    if env_name == 'DoubleHex_dir':
        env_name = 'Hex (Directed)'
    if env_name == 'double_large_maze_dir':
        env_name = 'Large Maze (Directed)'
    if env_name == 'four_rooms_dir':
        env_name = 'Four Rooms (Directed)'
    if env_name == 'double_large_maze':
        env_name = 'Large Maze'
    if env_name == 'rubiks':
        env_name = 'Rubiks Cube'
    if env_name == 'street':
        env_name = 'NYC (Directed)'
    return env_name
env_names = ['DoubleHex_dir','four_rooms_dir','torus','double_large_maze_dir','four_rooms_11_11.data','toh','office','DoubleHex','double_large_maze','rubiks']
option_names_and_num_options = [['NEO',4,0.1],['Cover',4,0.1],['Eigen',4,0.1],['NEO(first-ev)',1,0.1],['SP-novelty(most-novel)',1,0.1],['Qlearning-novelty',0,0.1]] 

beta_set = [0.5]
k_set = [64]
for beta in beta_set:
    option_names_and_num_options = []
    for k in k_set:
        for beta in beta_set:
            option_names_and_num_options.append(['hotspot_options' + str(beta) + '_' + str(k) + '_True', 4,0.1])
    env_names = ['DoubleHex_dir','four_rooms_dir','torus','double_large_maze_dir','four_rooms_11_11.data','toh','office','DoubleHex','double_large_maze','rubiks']
    for env in env_names:
            print(f"Environment: {env}")
            option_names = []
            alg = []
            for option_name, num_options, init_prob in option_names_and_num_options:
                save_name = 'NEO_n_' + option_name + '_False_True__' + str(num_options) + '_' + str(init_prob) + '.pkl'
                with open(save_name, "rb") as f:
                    results = pickle.load(f)
                if env not in results['env_results']:
                    continue
                print(f"Environment: {env}")
                print(f"Option Name: {option_name}")
                print(f"Num Options: {num_options}")
                print('number of nodes: ' , len(results['env_results'][env]['final_agent'].G.nodes()))
                print('number of edges: ' , len(results['env_results'][env]['final_agent'].G.edges()))
                if option_name == 'hotspot_options':
                    option_name = 'NEO'
                #print(np.shape(results['env_results'][env]['eval_returns']))
                alg.append(results['env_results'][env]['node_instance_list:'])
                #alg.append(results['env_results'][env]['edge_instance_list'])
                option_name = option_name.replace('hotspot_options', 'NEO (d=' + str(beta) + ' k=')
                option_name = option_name.replace('_', '') 
                option_name = option_name.replace('False', '') 
                option_name = option_name.replace('True', '') 
                option_names.append(option_name +')')
            #increase figure size
            plt.rcParams['figure.figsize'] = [6, 6]
            make_plots(alg, option_names, plot_title=re_title(env), cumulative=False, episodic=False, track_disc_reward=False, open_plot=False)
```
